```python
import pandas as pd
import numpy as np
df = pd.read_csv(r"C:\Users\marmh\Downloads\disney_plus_titles.csv")
```

```python
df['date_added'] = pd.to_datetime(df['date_added'])
```

```python
df['duration_num'] = df['duration'].str.extract('(\d+)').astype(float)
df['duration_type'] = df['duration'].str.extract('([a-zA-Z]+)')
df.fillna({'director': 'Unknown', 'cast': 'Unknown', 'country': 'Unknown', 'rating': 'Unknown'}, inplace=True)
```

```python
# This is where the actual project starts ⬇️
```

```python
df['description'] = df['description'].fillna('')
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['description'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)
```

```python
indices = pd.Series(df.index, index=df['title']).drop_duplicates()
```

```python
df['combined'] = df['description'] + " " + df['listed_in']
def recommend(title):
    idx = indices[title]

    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)[1:6]

    return df[['title', 'type', 'rating']].iloc[[i[0] for i in sim_scores]]
```

```python
recommended("Moana")
```

```python
recommended("Ernest Saves Christmas")
```
